```python
#HEATMAP GRAPG
plt.figure(figsize=(10,6))

sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
#SCATTER PLOT
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x="Sales", y="Quantity", color="green")
plt.title("SCATTER PLOT OF SALES VS QUANTITY")
plt.xlabel("Sales")
plt.ylabel("Quantity")
plt.show()
```

```python
###BAR GRAPH ##
colors = ["orange", "green", "blue", "red", "yellow"]
top_products=(
    df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(5)
)
plt.figure(figsize=(10,6))
top_products.plot(kind='bar', color=colors)
plt.title("Top 5 Products by Sales")
plt.xlabel("Product Name")
plt.ylabel("Total Sales")
plt.xticks(rotation=60)
plt.show()

```

```python
## HISTROGRAM PLOT###
plt.figure(figsize=(8,6))
sns.histplot(df["Region"],bins=80,color='red')
plt.title("DISTRIBUTION OF Region")
plt.xlabel("Region")
plt.ylabel("Frequency")
plt.show()
```

```python
# 3 PIE CHART FOR TOP 10 CITIES
top_cities = df['City'].value_counts().head(10)
plt.figure(figsize=(10,5))
plt.pie(top_cities, labels=top_cities.index, autopct='%1.1f%%')
plt.title("Top 10 Cities ")

plt.show()
```
