# Executive Summary

This project analyzes a retail sales dataset to understand customer purchasing behavior, product performance, and sales trends.

The analysis identifies the top-selling products, countries with the highest number of orders, and the relationship between product

 quantity and unit price.The findings can help businesses improve sales performance and make better business decisions.

```python
top_products = df["Description"].value_counts().head(10).reset_index()
top_products.columns = ["Product", "Count"]


colors = ["red","yellow","blue","black","brown",
          "green","skyblue","pink","purple","grey"]
plt.figure(figsize=(10,8))
sns.barplot(data=top_products, x="Product",y="Count",palette=colors,legend=False)

plt.title("Top 10 Products by Number of Orders")
plt.xlabel("Product")
plt.ylabel("Number of Orders")
plt.xticks(rotation=60, ha="right")

plt.show()
```

"This BarChart shows that the top 10 products with the highest number of orders.
 The WHITE HANGING HEART T-LIGHT HOLDER has the highest number of orders.

```python
## SCATTER PLOT ###
plt.figure(figsize=(12,6))
sns.scatterplot(data=df,x="UnitPrice",y="Quantity",color='green')
plt.title("Scatter plot of UnitOrice vs Quantity")
plt.xlabel("UnitPrice")
plt.ylabel("Quantity")

plt.show()
```

"This scatter plot shows that relationship between UnitPrice and Quantity.Most products have low prices
 and are purchase in small quantity.A few outliers indicate products with unusually high values.

```python
#BAR CHART ##
top_countries= df["Country"].value_counts().head(5).reset_index()

top_countries.columns = ["Country", "Count"]
plt.figure(figsize=(8,4))

sns.barplot(data=top_countries,x="Country",y="Count",color='red')

plt.title("Top 5 Countries by Number of Orders")
plt.xlabel("Country")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30)

plt.show()
```

"This BarChart shows that the top 5 Countries with the highest number of orders.
 The UNITED KINGDOM has the highest number of orders.

```python
## PIE CHART ##
top_countries = df['Country'].value_counts().head(6)
plt.figure(figsize=(10,6))
plt.pie(top_countries,labels=top_countries.index,autopct='%1.1f%%')
plt.title("Top 6 Countries by Number of orders")
plt.show()
```

"This pie chart represents the percentage of orders from the top 6 countries."

```python
## HIST0GRAM ##
plt.figure(figsize=(10,8))

sns.histplot(df["Description"], bins=40,color="skyblue")

plt.title("Distribution of Description")
plt.xlabel("Description")
plt.ylabel("Frequency")

plt.show()
```

Histrogram Graph shows that the distribution of product quantities in the dataset.
Most orders have low quantities.

Executive Findings
1. Top Products:-
The analysis identified the top 10 most frequently ordered products. These products have
 the highest customer demand and contribute significantly to overall sales.

2. Country Performance:-
The majority of orders were placed from the United Kingdom, making it the largest market 
in the dataset. Other countries contributed a much smaller share of total orders.

3. Unit Price vs Quantity:-

The scatter plot shows that most products with lower unit prices were purchased in
 higher quantities, while expensive products were generally ordered in smaller quantities.

4. Sales Distribution:-

The visualizations indicate that customer demand is concentrated on a limited number
 of products. This suggests that a small group of products generates a large portion of the business activity.

Business Recommendations
1. Maintain Stock Availability:-

Keep sufficient inventory for the top-selling products to avoid 
stock shortages and improve customer satisfaction.

2. Focus on High-Demand Markets:-

Since the United Kingdom has the highest number of orders, businesses should
 continue investing in marketing and customer service for this market.

3. Promote Low-Selling Products

Offer discounts, promotions, or product bundles to increase
 the sales of products with lower demand.

4. Optimize Pricing Strategy

Review product pricing regularly to maintain a balance between profitability
 and customer demand, especially for products with high sales volumes.

Conclusion

This analysis provides valuable insights into customer purchasing behaviour, product demand, and market performance. The findings can help businesses improve inventory management, pricing strategies, and sales planning, leading to better decision-making and increased business growth.
